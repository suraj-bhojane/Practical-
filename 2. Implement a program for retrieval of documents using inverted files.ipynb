```python
import re

documents = {
    "D1": "Artificial intelligence and machine learning",
    "D2": "Machine learning is branch of artificial intelligence",
    "D3": "Python is widely used for data science and machine learning",
}
```

```python
inverted_index = {}

for doc_id, text in documents.items():

    words = re.findall(r'\b\w+\b', text.lower())

    for word in words:
        if word not in inverted_index:
            inverted_index[word] = set()

        inverted_index[word].add(doc_id)
```

```python
print("INVERTED INDEX")
print("-" * 40)

for term in sorted(inverted_index):
    print(term, ":", sorted(inverted_index[term]))
```

```python
query = "python"
```

```python
if query in inverted_index:
    result = inverted_index[query]

    print("\nRETRIEVED DOCUMENTS")
    print("-" * 40)

    for doc_id in sorted(result):
        print(doc_id, ":", documents[doc_id])
else:
    print("\nNo matching documents found.")
```
