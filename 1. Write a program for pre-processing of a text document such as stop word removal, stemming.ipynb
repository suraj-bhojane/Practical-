```python
import nltk 
from nltk.tokenize import word_tokenize 
from nltk.corpus import stopwords 
from nltk.stem import PorterStemmer 
```

```python
text = """Information retrieval systems retrieve useful information from large collections of documents. 
          Retrieval systems process queries and documents efficiently.""" 
```

```python
tokens = word_tokenize(text.lower()) 
stop_words = set(stopwords.words("english"))
```

```python
filtered = [w for w in tokens if w.isalpha() and w not in stop_words] 
```

```python
stemmer = PorterStemmer() 
stems = [stemmer.stem(w) for w in filtered] 
```

```python
print("Original Text:", text)
print(f"Original tokens: {tokens}") 
print(f"\nAfter stop-word removal: {filtered}") 
print(f"\nAfter stemming: {stems}") 
print("\nVocabulary sizes") 
print(f"Before filtering: {len(set(w for w in tokens if w.isalpha()))}") 
print(f"After filtering : {len(set(filtered))}") 
print(f"After stemming  : {len(set(stems))}") 
```
