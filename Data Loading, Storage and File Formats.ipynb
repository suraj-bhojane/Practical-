# Practical 1: Data Loading, Storage and File Formats

Analyzing Sales Data from Multiple File Formats  
To load sales data from CSV, Excel and JSON files, inspect their structure, identify data quality issues, clean and transform the data, perform analysis, and visualize useful results.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_csv = pd.read_csv("dataset/sales_data_sample.csv", encoding="cp1252")

df_excel = pd.read_excel("dataset/Sales.xlsx")

df_json = pd.read_json("dataset/sales.json")

print("CSV shape:", df_csv.shape)
print("Excel shape:", df_excel.shape)
print("JSON shape:", df_json.shape)
```

```python
print("CSV DATA")
display(df_csv.head())

print("EXCEL DATA")
display(df_excel.head())

print("JSON DATA")
display(df_json.head())
```

```python
print("CSV information:")
df_csv.info()
print("\\nCSV statistical summary:")
display(df_csv.describe(include="all"))

print("\\nExcel information:")
df_excel.info()
print("\\nExcel statistical summary:")
display(df_excel.describe(include="all"))

print("\\nJSON information:")
df_json.info()
print("\\nJSON statistical summary:")
display(df_json.describe(include="all"))
```

```python
print("Missing values in CSV:")
display(df_csv.isnull().sum())

print("Duplicate rows in CSV:", df_csv.duplicated().sum())

print("\\nMissing values in Excel:")
display(df_excel.isnull().sum())

print("Duplicate rows in Excel:", df_excel.duplicated().sum())

print("\\nMissing values in JSON:")
display(df_json.isnull().sum())

print("Duplicate rows in JSON:", df_json.duplicated().sum())
```

```python
csv_clean = df_csv.copy()
excel_clean = df_excel.copy()
json_clean = df_json.copy()

csv_clean = csv_clean.drop_duplicates()
csv_clean["ORDERDATE"] = pd.to_datetime(csv_clean["ORDERDATE"], errors="coerce")
for col in ["QUANTITYORDERED", "PRICEEACH", "SALES", "MSRP"]:
    csv_clean[col] = pd.to_numeric(csv_clean[col], errors="coerce")

excel_clean = excel_clean.replace(["n.a.", "N.A.", "NA", "-1", -1], np.nan)
month_cols = ["JANUARY", "FEBRUARY", "MARCH", "APRIL", "MAY", "JUNE",
              "JULY", "AUGUST", "SEPTEMBER", "OCTOBER", "NOVEMBER", "DECEMBER"]
for col in month_cols:
    excel_clean[col] = pd.to_numeric(excel_clean[col], errors="coerce")
excel_clean = excel_clean.drop_duplicates()

json_clean = json_clean.drop_duplicates()
json_clean["Date"] = pd.to_datetime(json_clean["Date"], errors="coerce")
for col in ["Quantity", "Price", "Sales"]:
    json_clean[col] = pd.to_numeric(json_clean[col], errors="coerce")

print("Cleaning completed.")
print("CSV shape:", csv_clean.shape)
print("Excel shape:", excel_clean.shape)
print("JSON shape:", json_clean.shape)
```

```python
print("Missing values after cleaning - CSV:")
display(csv_clean.isnull().sum())

print("Missing values a" \
"fter cleaning - Excel:")
display(excel_clean.isnull().sum())

print("Missing values after cleaning - JSON:")
display(json_clean.isnull().sum())
```

```python
# Create a year-month column and calculate sales by product line
csv_clean["ORDER_MONTH"] = csv_clean["ORDERDATE"].dt.to_period("M").astype(str)

print("Total sales in CSV dataset:", round(csv_clean["SALES"].sum(), 2))

sales_by_product = csv_clean.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=False)
print("\\nSales by product line:")
display(sales_by_product)

sales_by_country = csv_clean.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False)
print("\\nTop 10 countries by sales:")
display(sales_by_country.head(10))
```

```python
excel_long = excel_clean.melt(
    id_vars=["SALES_ID", "SALES_BY_REGION"],
    value_vars=month_cols,
    var_name="MONTH",
    value_name="SALES"
)

print("Transformed Excel data:")
display(excel_long.head(10))

region_totals = excel_long.groupby("SALES_BY_REGION")["SALES"].sum().sort_values(ascending=False)
print("\\nTotal sales by region:")
display(region_totals)
```

```python
# Check whether Sales equals Quantity multiplied by Price
json_clean["Calculated_Sales"] = json_clean["Quantity"] * json_clean["Price"]

print("JSON order data:")
display(json_clean)

print("Total sales from JSON:", json_clean["Sales"].sum())
print("\\nSales by category:")
json_category_sales = json_clean.groupby("Category")["Sales"].sum().sort_values(ascending=False)
display(json_category_sales)
```

```python
csv_summary = (
    csv_clean.groupby("PRODUCTLINE", as_index=False)["SALES"].sum()
    .rename(columns={"PRODUCTLINE": "Group", "SALES": "Total_Sales"})
)
csv_summary["Source"] = "CSV - Product Line"

excel_summary = (
    region_totals.rename_axis("Group").reset_index(name="Total_Sales")
)
excel_summary["Source"] = "Excel - Region"

json_summary = (
    json_category_sales.rename_axis("Group").reset_index(name="Total_Sales")
)
json_summary["Source"] = "JSON - Category"

common_summary = pd.concat(
    [csv_summary, excel_summary, json_summary],
    ignore_index=True
)[["Source", "Group", "Total_Sales"]]

display(common_summary)
```

```python
sales_by_product.plot(kind="bar", figsize=(9, 5), title="Sales by Product Line (CSV)")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
json_category_sales.plot(kind="pie", autopct="%1.1f%%", figsize=(6, 6),
                         title="Sales Distribution by Category (JSON)")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
plt.boxplot(json_clean["Sales"].dropna())
plt.title("Distribution of Order Sales (JSON)")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
csv_clean.to_csv("cleaned_sales_csv.csv", index=False)
excel_clean.to_excel("cleaned_sales_excel.xlsx", index=False)
json_clean.to_json("cleaned_sales_json.json", orient="records", indent=4)
common_summary.to_csv("common_sales_summary.csv", index=False)

print("Cleaned datasets and common summary exported successfully.")
```
