```python
import networkx as nx
```

```python
G = nx.DiGraph()

G.add_edges_from([
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('C', 'A'),
    ('D', 'C')
])

print("Web Page Graph Created Successfully")
```

```python
pagerank = nx.pagerank(G)

print("PageRank Scores:")
for page, score in pagerank.items():
    print(page, ":", round(score, 4))
```

```python
ranked_pages = sorted(
    pagerank.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Web Pages Ranking:")

for page, score in ranked_pages:
    print(page, ":", round(score, 4))
```
